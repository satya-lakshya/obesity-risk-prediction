## Splitting into Training and Test set

```python
from sklearn.model_selection import train_test_split

X = df.drop(['NObeyesdad','id'], axis=1)
y = df['NObeyesdad']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)
```

```python
# Encoding categorical variables
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# ordinal, binary and nominal columns
ordinal_cols=['CAEC','CALC']
caec_order = ['no', 'Sometimes', 'Frequently', 'Always']
calc_order = ['no', 'Sometimes', 'Frequently']
ordinal_orders = [caec_order, calc_order]

nominal_cols=['MTRANS']

binary_cols = [
    'Gender', 
    'family_history_with_overweight', 
    'FAVC', 
    'SMOKE', 
    'SCC'
]

# creating a columntransformer
preprocessor=ColumnTransformer(
    transformers=[
        ('ordinal',OrdinalEncoder(categories=ordinal_orders,handle_unknown='use_encoded_value',unknown_value=-1),ordinal_cols),
        ('binary',OneHotEncoder(drop='first',sparse_output=False),binary_cols),
        ('nominal',OneHotEncoder(handle_unknown='ignore',sparse_output=False),nominal_cols),
        ('num',StandardScaler(),numerical_cols)
    ],
    remainder='passthrough'
)

# Apply preprocessing to X_train and X_test
preprocessor.fit(X_train)

feature_names = preprocessor.get_feature_names_out()

X_train_processed = pd.DataFrame(
    preprocessor.transform(X_train),
    columns=feature_names,
    index=X_train.index
)   

X_test_processed = pd.DataFrame(
    preprocessor.transform(X_test),
    columns=feature_names,
    index=X_test.index,
)

# defining target variable order
nobeyesdad_order = [
    'Insufficient_Weight', 
    'Normal_Weight', 
    'Overweight_Level_I', 
    'Overweight_Level_II', 
    'Obesity_Type_I', 
    'Obesity_Type_II', 
    'Obesity_Type_III'
]
#  encoding target column
y_order = [nobeyesdad_order]
y_encoder = OrdinalEncoder(categories=y_order)
y_train_processed = pd.Series(
    y_encoder.fit_transform(y_train.values.reshape(-1, 1)).ravel(),
    index=y_train.index,
    name='NObeyesdad'
)

y_test_processed = pd.Series(
    y_encoder.transform(y_test.values.reshape(-1, 1)).ravel(),
    index=y_test.index,
    name='NObeyesdad'
)





```

## Training Models - Building Baseline

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report,accuracy_score, confusion_matrix

# Decision Tree
dt_clf = DecisionTreeClassifier(random_state=42)
dt_clf.fit(X_train_processed, y_train_processed)

# Logistic Regression
lr_clf = LogisticRegression(max_iter=1000, random_state=42)
lr_clf.fit(X_train_processed, y_train_processed)

# K-Nearest Neighbors
knn_clf = KNeighborsClassifier(n_neighbors=5)
knn_clf.fit(X_train_processed, y_train_processed)

# defining a function to generate predictions, compute accuracy, give classification report and plot confusion matrix
def predict_and_plot(model,
                     inputs,
                     targets,
                     name_data='',
                     name_model=''):
    
    print(f"{name_data} Data")
    print(f"{name_model}'s Report:")
    
    preds=model.predict(inputs)
    accuracy=accuracy_score(targets,preds)
    print(f"Accuracy : {accuracy*100:.2f}%")
    
    print('Classification Report:', classification_report(targets, preds))

    cf=confusion_matrix(targets,preds,normalize='true')
    plt.figure()
    sns.heatmap(cf,annot=True,xticklabels=nobeyesdad_order,yticklabels=nobeyesdad_order)
    plt.xlabel("Prediction")
    plt.ylabel('Target')
    plt.title(f"{name_model}-{name_data} Set - Confusion Matrix")

    return preds

    
    

```

```python
y_train_pred_dt = predict_and_plot(dt_clf,
                                 X_train_processed,
                                 y_train_processed,
                                 "Train",
                                 "Decision Tree")
```

```python
y_train_pred_lr = predict_and_plot(lr_clf,
                                 X_train_processed,
                                 y_train_processed,
                                 "Train",
                                 "Logistic Regression")
```

```python
y_train_pred_knn = predict_and_plot(knn_clf,
                                 X_train_processed,
                                 y_train_processed,
                                 "Train",
                                 "KNN")
```

## Hyperparameter Tuning

```python
dt_clf.get_params()
```

```python
from sklearn.model_selection import GridSearchCV

def tune_model(model, param_grid, X_train, y_train, cv=5,show_plots=True):
    
    grid_search = GridSearchCV(estimator=model,
                               param_grid=param_grid,
                               cv=cv,
                               scoring='accuracy',
                               return_train_score=True,
                               n_jobs=-1)  
    grid_search.fit(X_train, y_train)
    
    results_df = pd.DataFrame(grid_search.cv_results_)

    if show_plots==True:
        for key in list(param_grid.keys()) :
            plt.figure(figsize=(10,6))
            plt.title(f'Overfitting Curve of {key}:')
            plt.plot(results_df[f'param_{key}'],results_df['mean_train_score'],'b-o')
            plt.plot(results_df[f'param_{key}'],results_df['mean_test_score'],'r-o')
            plt.xlabel(key)
            plt.ylabel('accuracy')
            plt.legend(['train_score','test_score'])
    
    print("Best Parameters:", grid_search.best_params_)
    print("Best CV Accuracy:", grid_search.best_score_)

    
    return grid_search
```

```python
# tuning decision tree
param_grid_dt = {
    'max_depth': [x for x in range(1,28)],
}
results_dt=tune_model(dt_clf,param_grid_dt,X_train_processed,y_train_processed)
results_dt_df=pd.DataFrame(results_dt.cv_results_)
best_dt_clf=results_dt.best_estimator_
```

```python
lr_clf.get_params()
```

```python
# tuning linear regression model
import numpy as np
param_grid_lr={
    'C':[1,2,3,4,4.5,4.8,5,5.1,5.5]
}
results_lr=tune_model(lr_clf,param_grid_lr,X_train_processed,y_train_processed)
results_lr_df=pd.DataFrame(results_lr.cv_results_)
best_lr_clf=results_lr.best_estimator_
```

```python
knn_clf.get_params()
```

```python
param_grid_knn = {
    'n_neighbors': [x for x in range(1,20)]
}

results_knn=tune_model(knn_clf,param_grid_knn,X_train_processed,y_train_processed)
results_knn_df=pd.DataFrame(results_knn.cv_results_)
best_knn_clf=results_knn.best_estimator_
```

```python
# COMPARING THE 3 MODELS ON TEST SETS
print("Decision Tree Accuracy:", accuracy_score(y_test_processed, best_dt_clf.predict(X_test_processed)))
print("Logistic Regression Accuracy:", accuracy_score(y_test_processed, best_lr_clf.predict(X_test_processed)))
print("KNN Accuracy:", accuracy_score(y_test_processed, best_knn_clf.predict(X_test_processed)))
```
